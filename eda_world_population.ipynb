```python
# Visualize missing values
plt.figure(figsize=(12,6))
sns.heatmap(df.isna(), cbar=False, yticklabels=False, cmap='viridis')
plt.title("Missing Values Heatmap")
plt.show()
```

```python
pop_cols = ['2022 Population', '2020 Population', '2015 Population', '2010 Population',
            '2000 Population', '1990 Population', '1980 Population', '1970 Population']

other_numeric = ['Area (km²)', 'Density (per km²)', 'Growth Rate']
```

```python
# Handle population missing values using linear interpolation
df[pop_cols] = df[pop_cols].interpolate(method='linear', axis=1, limit_direction='both')
```

```python
# Handle other numeric columns using continent-wise median
for col in other_numeric:
    df[col] = df.groupby('Continent')[col].transform(lambda x: x.fillna(x.median()))
```

```python
print("\n=== Missing Values After ===")
print(df.isna().sum())
```

```python
# 6. Summary Statistics
print("\n=== Population Summary Statistics ===")
pop_cols = [col for col in df.columns if 'population' in col.lower() or col.lower() in ['pop2020', 'pop2010']]
print(df[pop_cols].describe())
```

```python
# 7. Explore categorical variables
print("\n=== Continent Counts ===")
print(df['Continent'].value_counts())

plt.figure(figsize=(10,5))
sns.countplot(data=df, x='Continent', palette='Set2')
plt.title("Number of Countries per Continent")
plt.xticks(rotation=45)
plt.show()
```

```python
# 8. Visualize Population Distribution
# Total population column assumption: last column or named 'population_total'
if 'population_total' in df.columns:
    plt.figure(figsize=(12,6))
    sns.histplot(df['population_total'], bins=50, kde=True)
    plt.title("Distribution of Total Population")
    plt.xlabel("Population")
    plt.show()
else:
    # Plot 2020 population as example
    plt.figure(figsize=(12,6))
    sns.histplot(df[df.columns[-4]], bins=50, kde=True)
    plt.title(f"Distribution of {df.columns[-4]}")
    plt.xlabel("Population")
    plt.show()
```

```python
# 9. Correlation analysis
numeric_cols = ['2022 Population', '2020 Population', '2015 Population', '2010 Population',
                '2000 Population', '1990 Population', '1980 Population', '1970 Population',
                'Area (km²)', 'Density (per km²)', 'Growth Rate', 'World Population Percentage']

plt.figure(figsize=(12,10))
corr = df[numeric_cols].corr()
sns.heatmap(corr, annot=True, fmt=".2f", cmap='coolwarm', cbar=True)
plt.title("Correlation Matrix")
plt.show()
```

by looking at the correlation table we can get some intakes.
E.g  2022 population is very higly correlated to previous year population, because for most countries population is increaing. 
But if we look at area or density, it is not higly correlated. Reason could be there are some countries with high population but area is small. or vice verse

as it does not look good, let's change the size

```python
# 10. Population Trends Over Years
# Assuming columns from 2010 to 2020 are sequentially population data
pop_years = df.columns[6:13]  # adjust depending on your dataset
df_trend = df[['Country'] + list(pop_years)].set_index('Country').T
df_trend.plot(figsize=(15,6), legend=False)
plt.title("Population Trend per Country (subset)")
plt.xlabel("Year")
plt.ylabel("Population")
plt.show()
```

```python
# 11. Top 10 Most Populous Countries
df['latest_population'] = df[pop_years[-1]]
top10 = df[['Country', 'latest_population']].sort_values(by='latest_population', ascending=False).head(10)

plt.figure(figsize=(12,6))
sns.barplot(data=top10, x='Country', y='latest_population', palette='magma')
plt.title("Top 10 Most Populous Countries")
plt.xticks(rotation=45)
plt.show()
```

```python
#12. Scatter Plot Examples
# Population vs Area if 'area' column exists
if 'area' in df.columns:
    plt.figure(figsize=(12,6))
    sns.scatterplot(data=df, x='area', y='latest_population', hue='region', palette='Set1', s=100)
    plt.title("Population vs Area by Region")
    plt.xscale('log')
    plt.yscale('log')
    plt.show()
```
